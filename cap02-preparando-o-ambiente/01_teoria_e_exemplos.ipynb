# 💡 2. Primeiros Programas e Conceitos Matemáticos

## 💻 2. Primeiros programas em Python

O Python pode ser utilizado diretamente como uma calculadora para realizar operações matemáticas básicas e avançadas.

### ➕ Operações Básicas

- ➕ **Adição:** utiliza o sinal de mais (`+`).

```python
print(2 + 3)
```

* ➖ **Subtração:** utiliza o sinal de menos (`-`).

```python
print(5 - 3)
```

* 🔄 **Combinação de operações:** adição e subtração podem ser combinadas na mesma linha.

```python
print(10 - 4 + 2)
```

* ✖️ **Multiplicação:** representada por um **asterisco (`*`)**.

```python
print(2 * 10)
```

* ➗ **Divisão comum:** representada por uma **barra (`/`)**, podendo produzir resultados inteiros ou fracionários.

```python
print(20 / 4)  # Resultado exato
print(10 / 4)  # Resultado fracionário (2.5)
```

* 🔢 **Divisão inteira:** representada por **duas barras (`//`)**, retorna apenas a parte inteira do resultado.

```python
print(10 // 4)  # Resultado: 2
```

* 🚀 **Exponenciação (potenciação):** representada por **dois asteriscos (`**`)**, eleva um número a um expoente.

```python
print(2 ** 3)  # Resultado: 8
```

* 🧩 **Módulo (resto da divisão):** representado pelo símbolo de porcentagem (`%`), retorna o que sobra de uma divisão.

```python
print(10 % 3)  # Resultado: 1
print(16 % 7)  # Resultado: 2
print(63 % 8)  # Resultado: 7
```

### 💡 Lembrete: Exponenciação e Raízes

Na expressão $2^{3}$ (que equivale a $2 \times 2 \times 2$), o **2** é a base e o **3** é o expoente. Lê-se "2 elevado à potência de 3" ou "2 elevado a 3".

* 🔹 **Propriedades fundamentais da potenciação:**
* Todo número elevado a zero é igual a 1: $x^{0} = 1$
* Todo número elevado a 1 é igual a si mesmo: $x^{1} = x$
* O número 1 elevado a qualquer expoente é sempre 1: $1^{x} = 1$
* O número 0 elevado a qualquer número positivo é igual a 0: $0^{x} = 0$


* 🌱 **Cálculo de raízes:** podemos calcular a raiz de um número usando exponenciação com expoente fracionário e positivo. A fórmula $a^{\frac{x}{y}}$ equivale a $\sqrt[y]{a^{x}}$. Por exemplo, $\sqrt[3]{2}$ equivale a $2^{\frac{1}{3}}$:

```python
print(2 ** (1 / 3))
```

### 📊 Resumo de Operadores Matemáticos

| Operador | Operação |
| --- | --- |
| `+` | Adição |
| `-` | Subtração |
| `*` | Multiplicação |
| `/` | Divisão (com resultado fracionário) |
| `//` | Divisão (com resultado inteiro) |
| `%` | Módulo ou resto da divisão |
| `**` | Exponenciação ou potenciação |

### ⚖️ Ordem de Precedência das Operações

Assim como na matemática tradicional, os **parênteses `()**` são usados em Python para alterar a ordem de execução de uma expressão. Sem parênteses, a ordem de prioridade é:

1. **Exponenciação ou potenciação (`**`)**
2. **Multiplicação (`*`), divisão (`/` e `//`) e módulo (`%`)**
3. **Adição (`+`) e subtração (`-`)**

Por exemplo, a expressão Python `1500 + (1500 * 5 / 100)` equivale matematicamente a:

$$1500 + \left(\frac{1500 \times 5}{100}\right)$$

## 🧮 3. Revisão de conceitos matemáticos

Abaixo estão os principais conceitos e propriedades matemáticas fundamentais para a programação.

### 🔢 3.1 Conjuntos Numéricos

* 🌱 **Números naturais:** números inteiros não negativos. Ex.: $0, 1, 2, 3, \dots$
* 🌡️ **Números inteiros:** números naturais e seus opostos negativos. Ex.: $\dots, -3, -2, -1, 0, 1, 2, 3, \dots$
* 🍰 **Números racionais:** números que podem ser escritos como fração de números inteiros. Incluem os naturais, inteiros, decimais finitos e dízimas periódicas. Ex.: $-4$; $5$; $1,5$; $3,9$; $1 / 3 = 0,3333333\dots$
* 🌀 **Números irracionais:** números com casas decimais infinitas e não periódicas, que não podem ser representados por frações. Ex.: $\pi \approx 3,141592\dots$
* 🌍 **Números reais:** conjunto formado pela união de todos os números racionais e irracionais.

### ➕➖ 3.2 Propriedades da adição e da subtração

* 🔄 **Operações inversas:** adição e subtração desfazem uma à outra.

$$A + B - B = A \quad \text{ou} \quad B - A + A = B$$


* 🔀 **Comutatividade na adição:** na soma, a ordem dos valores não altera o resultado.

$$A + B = B + A$$


* ⚠️ **Ordem na subtração:** na subtração, a ordem altera o resultado.

$$A - B \neq B - A$$


* 0️⃣ **Elemento neutro:** somar ou subtrair zero não altera o valor original.

$$A + 0 = A - 0 = A$$


* ➖ **Subtração como soma de opostos:** subtrair um número equivale a somar o seu valor negativo.

$$A + (-B) = A - B$$

### ➗ 3.3 Propriedades da divisão

Na operação $\text{Dividendo} / \text{Divisor} = \text{Quociente}$:

* **Dividendo:** o número que está sendo dividido.
* **Divisor:** o número que divide o dividendo.
* **Quociente:** o resultado da divisão. Na divisão inteira (`//`), o quociente é inteiro. Na divisão comum (`/`), pode ser inteiro ou fracionário.
* **Resto:** o valor que sobra quando a divisão não é exata, calculado por:

$$\text{Resto} = \text{Dividendo} - (\text{Quociente} \times \text{Divisor})$$

```python
print(101 // 4)  # Divisão inteira (Quociente: 25)
print(101 / 4)   # Divisão comum (Quociente: 25.25)
print(101 % 4)   # Módulo ou resto (Resto: 1)
```

*(Nota: em Python, devido às limitações de precisão do ponto flutuante, nem sempre é possível reverter exatamente uma divisão com números decimais).*

* 🎯 **Divisão exata e divisibilidade:** uma divisão é exata quando o resto é **zero** e o quociente é **inteiro**. Se `A % B == 0`, dizemos que $A$ é divisível por $B$ e que $A$ é múltiplo de $B$.

```python
print(10 % 3)  # Resto 1 (10 não é divisível por 3)
print(9 % 3)   # Resto 0 (9 é divisível por 3)
```

* 🍕 **Frações e simplificações:** uma fração $\frac{a}{b}$ representa a divisão do numerador $a$ pelo denominador $b$. Duas divisões são equivalentes quando têm o mesmo quociente e resto (ex.: $100 / 10$ e $10 / 1$ resultam em $10$). Uma fração é **irredutível** quando numerador e denominador não têm divisores comuns.
* Quando o denominador é $1$, a fração equivale ao próprio número inteiro: $\frac{4}{1} = 4$.
* Quando o numerador é menor que o denominador, o quociente fica entre $-1$ e $1$ (ex.: $\frac{2}{5} = 0,4$).


* 🚦 **Regra de sinais na divisão:** sinais iguais geram quociente **positivo**; sinais diferentes geram quociente **negativo**.

```python
print(-5 / -2)  # Resultado positivo (2.5)
print(5 / 2)    # Resultado positivo (2.5)
print(5 / -2)   # Resultado negativo (-2.5)
print(-5 / 2)   # Resultado negativo (-2.5)
```

* 🚫 **Divisão por zero:** matematicamente indeterminada. Em Python, gera uma exceção (`ZeroDivisionError`), que pode ser tratada com `try/except`:

```python
try:
    10 / 0
except ZeroDivisionError as e:
    print(f'Erro: ZeroDivisionError - {e}')
```

* 🔍 **Atenção com decimais:** divisões envolvendo números reais (ponto flutuante) exigem cuidado na interpretação dos resultados:

```python
print(2 / 5)
print(0.6 / 5)
print(5 / 0.6)
print(10 / 0.1)
```

### ✖️ 3.4 Propriedades da multiplicação

A multiplicação representa a soma sucessiva de um número. Em $A \times B$, $A$ é o **multiplicando**, $B$ é o **multiplicador** (ambos chamados de **fatores**) e o resultado é o **produto**.

$$4 \times 5 = 4 + 4 + 4 + 4 + 4 = 5 + 5 + 5 + 5 = 20$$

* 0️⃣ **Multiplicação por zero:** o produto é sempre $0$, independentemente da quantidade de fatores ($A \times B \times C \times 0 = 0$).
* 1️⃣ **Elemento neutro (multiplicação por 1):** o produto é igual ao próprio número ($A \times 1 = 1 \times A = A$).
* 🔄 **Inversão de sinal (multiplicação por -1):** inverte o sinal do número ($A \times -1 = -A$).

```python
print(5 * -1)     # -5
print(-5 * -1)    # 5
print(-3.4 * -1)  # 3.4
```

* 🔀 **Comutatividade:** a ordem dos fatores não altera o produto ($A \times B = B \times A$).

```python
print(6 * 2 == 2 * 6)         # True (12)
print(-4 * 3 == 3 * -4)       # True (-12)
print(10.4 * -8 == -8 * 10.4) # True (-83.2)
```

* 🚦 **Regra de sinais:** fatores com sinais diferentes geram produto negativo. Em multiplicações com vários fatores negativos, se a quantidade de fatores negativos for **par**, o resultado é **positivo**; se for **ímpar**, o resultado é **negativo**.

```python
print(1 * -1 * -1)    # 1 (2 fatores negativos -> positivo)
print(-1 * -1 * -1)   # -1 (3 fatores negativos -> negativo)
print(3 * -2 * 9)     # -54
print(-3 * -2 * 9)    # 54
print(-3 * -2 * -9)   # -54
```

* 🔢 **Números pares e ímpares:** um número é **par** se for $0$ ou múltiplo de $2$, representado pela fórmula $2k$ (onde $k$ é inteiro). Todos os demais são **ímpares**, representados por $(2k) + 1$.

### 🧩 3.5 Propriedades do resto

* 🔁 **Cálculo de limites e ciclos:** o operador módulo (`%`) é muito útil para criar sequências cíclicas que não ultrapassam um limite máximo. Por exemplo, usar `% 4` gera repetidamente a sequência `0, 1, 2, 3`:

```python
print(0 % 4)  # 0
print(1 % 4)  # 1
print(2 % 4)  # 2
print(3 % 4)  # 3
print(4 % 4)  # 0
print(5 % 4)  # 1
```

* 📉 **Módulo com números negativos:** em Python, quando usamos `%` com números negativos, o quociente da divisão é arredondado para baixo (em direção ao infinito negativo, função piso).
* Por exemplo, em `-10 % 9`, a divisão `-10 / 9` resulta em `-1,111...`, que é arredondada para `-2`.
* Aplicando a fórmula geral $\text{Módulo} = a - n \left\lfloor \frac{a}{n} \right\rfloor$:

$$\text{Módulo} = -10 - 9 \times (-2) = -10 + 18 = 8$$

```python
print(0.5 % 9)   # 0.5
print(11.5 % 9)  # 2.5
print(-10 % 9)   # 8
print(10 % 9)    # 1
print(-5 % 9)    # 4
```

### ⚖️ 3.6 Regra de 3 e proporção

A regra de 3 resolve problemas onde a proporção entre dois valores precisa ser mantida ($\frac{a}{b} = \frac{c}{d}$) por meio da multiplicação cruzada ($a \times d = b \times c$).

* 🔍 **Exemplo prático:** sabendo que $\frac{10}{2} = \frac{c}{3}$, multiplicamos cruzado para encontrar $c$:

$$2 \times c = 10 \times 3 \implies 2c = 30 \implies c = \frac{30}{2} = 15$$


* 📐 **Fórmulas diretas para encontrar cada variável:**
* $a = (b \times c) \div d$
* $b = (a \times d) \div c$
* $c = (a \times d) \div b$
* $d = (b \times c) \div a$

### 💯 3.7 Porcentagem

A porcentagem expressa uma proporção em relação a $100$. Assim, $10\%$ equivale à fração $\frac{10}{100}$.

* 📊 **Exemplo 1 (calcular o valor de uma porcentagem):** quanto é $10\%$ de $250$?

$$\frac{x}{250} = \frac{10}{100} \implies x = \frac{250 \times 10}{100} = 25$$


* 📈 **Exemplo 2 (descobrir o percentual):** $75$ representa quantos por cento de $450$?

$$\frac{75}{450} = \frac{x}{100} \implies x = \frac{75 \times 100}{450} \approx 16,66\%$$


* ⚠️ **Atenção à notação decimal:**
* $30\% = 30 \div 100 = 0,3$
* $0,3\% = 0,3 \div 100 = 0,003$ (não confunda com $3 \div 100$)

### 📊 3.8 Média aritmética

A média aritmética simples indica o valor central de um conjunto de números, somando todos os elementos e dividindo pela quantidade total. Para a sequência $2, 4, 5$:

$$\text{Média} = \frac{2 + 4 + 5}{3} = \frac{11}{3} \approx 3,666$$

```python
print((2 + 4 + 5) / 3)
```

### 🏋️ 3.9 Média ponderada

Na média ponderada, cada valor possui um peso diferente. Multiplica-se cada nota pelo seu respectivo peso e divide-se o resultado pela soma de todos os pesos.

* 📝 **Exemplo 1 (pesos inteiros):** testes com peso $2$ (nota $7$) e prova com peso $1$ (nota $9$). A soma dos pesos é $3$:

$$\text{Média} = \frac{(2 \times 7) + (1 \times 9)}{3} = \frac{14 + 9}{3} = \frac{23}{3} \approx 7,66$$

```python
print((2 * 7 + 1 * 9) / 3)
```

* 🎯 **Exemplo 2 (pesos percentuais):** testes valendo $20\%$ (nota $6$), trabalhos valendo $30\%$ (nota $8$) e prova valendo $50\%$ (nota $7$). A soma dos pesos é $100$:

$$\text{Média} = \frac{(6 \times 20) + (8 \times 30) + (7 \times 50)}{100} = 7,1$$

```python
print((6 * 20 + 8 * 30 + 7 * 50) / 100)
```

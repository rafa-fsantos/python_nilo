#### LEMBRETE

A exponenciação é uma operação importante, vale lembrar que $2^{3}$ é igual a $2 \times 2 \times 2$. No caso de $2^{3}$, 2 é a base e 3 o expoente, então lemos: "2 é elevado a potência de 3" ou simplesmente "2 elevado a 3".

A exponenciação tem também várias propriedades, como:

- Todo número elevado a zero é igual a 1: $x^{0} = 1$
- Todo número elevado a 1 é igual a sim mesmo: $x^{1} = x$
- Um elevado a não importa qual número é 1: $1^{x} = 1$
- Zero elevado a qualquer número positivo é igual a 0: $0^{x} = 0$

Podemos também calcular a raiz de um número pela exponenciação com expoente fracionário (e positivo). Dessa forma, $a^{\frac{x}{y}}$ é igual a $\sqrt[y]{a^{x}}$, ou seja, $2^{\frac{1}{3}}$ é igual a $\sqrt[3]{2}$, e em Python:

```python
2 ** (1 / 3)
```

Podemos também obter o resto (módulo) da divisão de dois números usando o símbolo **%**:

```python
10 % 3
```

```python
16 % 7
```

```python
63 % 8
```

**Operadores e operações matemáticas**

| Operador | Operação 
| :---: | :---
| + | Adição
| - | Subtração
| * | Multiplicação
| / | Divisão (com resultado fracionário)
| // | Divisão (com resultados inteiros)
| % | Módulo ou resto
| ** | Exponenciação ou potenciação

Os parênteses são utilizados em Python da mesma forma que em expressões matemáticas, ou seja, para alterar a ordem de execução de uma operação.

Vamos rever a ordem de precedência das operações:

1. Exponenciação ou potenciação (**)
2. Multiplicação (*), divisão (/ e //) e módulo (%)
3. Adição (+) e subtração (-)

Assim, a expressão  $1500 + (1500 * 5 / 100)$ é equivalente a:

$$1500 + (\frac{1500 \times 5}{100})$$

A **atribuição (=)** é o armazenamento de um valor em uma variável a qual armazenamos na memória do computador:

```python
# Programa 2.1: Primeiro programa com variáveis
a = 2
b = 3
print(a + b)
```

A função **print** realiza a impressão do resultado da adição das variáveis **a** e **b**, (parâmetros de print), assim como na matemática em que $f(x)$, em que $f$ é o nome da função, e $x$ um parâmetro.

Poderimaos obter o mesmo resultado de diversas formas:

```python
print(2 + 3)
```

Ou assim:

```python
print(5)
```

Separar valores de entrada, processamento e resultado é uma boa prática e facilita a legibilidade do programa

Ao resolvermos o problema com o uso de variáveis estamos dando nomes aos valores de entrada e facilitando o entendimento e alterações futuras.

#### TRIVIA

$a$ e $b$ são nomes horríveis para variáveis, mas servem de exemplo, pois não precisamos explicar o que é $a$ e o que é $b$, talvez os lados de um triângulo, mas para o exemplo não é importante. Quando você estiver escrevendo seus programas, tente
dar nomes significativos as suas variáveis. O você do futuro e todas as pessoas que lerão seus programas agradecerão.

```python
# Programa 2.2: Cálculo de aumento de salário
salario = 1500
aumento = 5
print(salario + (salario * aumento / 100))
```

O programa 2.2 pode ser escrito de forma mais direta, sem o uso de variáveis:

```python
# Programa 2.3: Alternativa para o cálculo de aumento de salário
print(1500 + (1500 * 5 / 100))
```

O uso de variáveis torna o o problema de forma genérica, separando valores de entrada e cálculo além disso, podemos referenciar o mesmo valor várias vezes com uma simples alteração.

No **Programa 2.3** é necessário alterar a mesmo dado em duas posições diferentes o que pode ocasionar erros e dificulta o entendimento do problema.

### 3. Revisão de conceitos matemáticos

Segue um resumo dos conceitos e propriedades das principais operações matemáticas:

#### 3.1 Números

Existem vários conjuntos numéricos:

- **Números naturais:** números inteiros maiores que zero - $0, 1, 2, 3, ...$
- **Números inteiros:** números naturais e seus opostos (com sinal negativo) - $...-3, -2, -1, 0, 1, 2, 3...$
- **Números racionais:** números que podme ser representados sob forma de fração de números inteiros - $-4, 5, 1.5, 3.9$.
    Além dos números naturais e inteiros, incluem-se nos racionais os todos os números com casas decimais finitas e dízimas periódicas - $1 / 3 = 0.3333333$ em que o 3 se repete infinitamente.
- **Números irracionais:** números com casas decimais infinitas e que não se repetem, portanto não podem ser representados por frações de números irredutíveis - $pi, 3.14151692$
- **Números reais:** conjunto formado pela união dos conjuntos do números racionais e irracionais.
